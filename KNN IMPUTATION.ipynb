```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.impute import KNNImputer,SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df=pd.read_csv('Titanic.csv')[['Survived','Fare','Age','Pclass']]
```

```python
df
```

```python
df.head()
```

```python
df.isnull().mean()*100
```

```python
x=df.drop(columns=['Survived'])
y=df['Survived']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train
```

```python
knn=KNNImputer()   # inside the knnimputer the paraeter called nabours if we change it they will change our value
x_train_trf=knn.fit_transform(x_train)
x_test_trf=knn.fit_transform(x_test)
```

```python
pd.DataFrame(x_train_trf,columns=x_train.columns)
```

```python
lr=LogisticRegression()
lr.fit(x_train_trf,y_train)
y_pred=lr.predict(x_test_trf)     # apply the logitic regreesion on knn impute data
accuracy_score(y_pred,y_test)     
```

```python
knn=KNNImputer(weights='distance')   
x_train_trf=knn.fit_transform(x_train)     # now we used the parmeter weights=ditanse they imprve our accuarcy from privious one
x_test_trf=knn.fit_transform(x_test)
pd.DataFrame(x_train_trf,columns=x_train.columns)
```

```python
lr=LogisticRegression()
lr.fit(x_train_trf,y_train)
y_pred=lr.predict(x_test_trf)     
accuracy_score(y_pred,y_test)     
```

```python
# now we will try simpleimputer mean t cheak is acuracy is increse or not
si=SimpleImputer()
x_train_trf2=si.fit_transform(x_train)
x_test_trf2=si.transform(x_test)
```

```python
lr=LogisticRegression()
lr.fit(x_train_trf2,y_train)
y_pred2=lr.predict(x_test_trf2)
accuracy_score(y_pred2,y_test)   # see the accuracy f simple imputer which is less than knn imputer
```
